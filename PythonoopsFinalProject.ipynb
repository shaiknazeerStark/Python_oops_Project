```python

# 1. VEHICLE CLASS

class Vehicle:
    def __init__(self,vid,brand,model,rate):
        self.vid=vid
        self.brand=brand
        self.model=model
        self.__rate=rate
        self.__available=True

    def get_rental_rate(self):
        return self.__rate

    def check_availability(self):
        return self.__available

    def change_availability(self,status):
        self.__available=status

    def calculate_cost(self,days):
        return self.__rate*days

    def display_details(self):
        print("Vehicle ID:",self.vid)
        print("Brand:",self.brand)
        print("Model:",self.model)
        print("Rate per day:",self.__rate)
        print("Available:",self.__available)


# 2. CAR CLASS

class Car(Vehicle):
    def __init__(self,vid,brand,model,rate,seats):
        super().__init__(vid,brand,model,rate)
        self.seats=seats

    def calculate_cost(self,days):
        cost=super().calculate_cost(days)
        cost=cost+(200*days)
        return cost

    def display_details(self):
        super().display_details()
        print("Seats:",self.seats)


# 3. BIKE CLASS

class Bike(Vehicle):
    def __init__(self,vid,brand,model,rate,cc):
        super().__init__(vid,brand,model,rate)
        self.cc=cc

    def calculate_cost(self,days):
        cost=super().calculate_cost(days)
        cost=cost+(50*days)
        return cost

    def display_details(self):
        super().display_details()
        print("Engine CC:",self.cc)


# 4. PREMIUM CAR CLASS

class PremiumCar(Car):
    def __init__(self,vid,brand,model,rate,seats,feature):
        super().__init__(vid,brand,model,rate,seats)
        self.feature=feature

    def calculate_cost(self,days):
        cost=super().calculate_cost(days)
        cost=cost+(500*days)
        return cost

    def display_details(self):
        super().display_details()
        print("Premium feature:",self.feature)


# 5. CUSTOMER CLASS

class Customer:
    def __init__(self,cid,name,phone,licence):
        self.cid=cid
        self.name=name
        self.phone=phone
        self.licence=licence

    def display_details(self):
        print("Customer ID:",self.cid)
        print("Name:",self.name)
        print("Phone:",self.phone)
        print("Licence:",self.licence)


# 6. RENTAL CLASS

class Rental:
    def __init__(self,customer,vehicle,days,amount):
        self.customer=customer
        self.vehicle=vehicle
        self.days=days
        self.amount=amount
        self.status="Active"

    def display_rental(self):
        print("Customer:",self.customer.name)
        print("Vehicle:",self.vehicle.brand,self.vehicle.model)
        print("Days:",self.days)
        print("Amount:",self.amount)
        print("Status:",self.status)


# 7. RENTAL AGENCY CLASS

class RentalAgency:
    def __init__(self):
        self.vehicles=[]
        self.customers=[]
        self.rentals=[]

    def add_vehicle(self,vehicle):
        for v in self.vehicles:
            if v.vid==vehicle.vid:
                print("Vehicle ID already exists")
                return

        self.vehicles.append(vehicle)
        print("Vehicle added successfully")

    def add_customer(self,customer):
        for c in self.customers:
            if c.cid==customer.cid:
                print("Customer ID already exists")
                return

        self.customers.append(customer)
        print("Customer registered successfully")

    def find_vehicle(self,vid):
        for v in self.vehicles:
            if v.vid==vid:
                return v
        return None

    def find_customer(self,cid):
        for c in self.customers:
            if c.cid==cid:
                return c
        return None

    def show_vehicles(self):
        for v in self.vehicles:
            print("--------------------")
            v.display_details()

    def rent_vehicle(self,cid,vid,days):
        if days<=0:
            print("Days must be greater than zero")
            return

        customer=self.find_customer(cid)
        vehicle=self.find_vehicle(vid)

        if customer is None:
            print("Customer not found")
        elif vehicle is None:
            print("Vehicle not found")
        elif not vehicle.check_availability():
            print("Vehicle is already rented")
        else:
            amount=vehicle.calculate_cost(days)

            rental=Rental(customer,vehicle,days,amount)
            self.rentals.append(rental)

            vehicle.change_availability(False)

            print("Vehicle rented successfully")
            print("Rental amount:",amount)

    def return_vehicle(self,vid):
        for r in self.rentals:
            if r.vehicle.vid==vid and r.status=="Active":
                r.status="Returned"
                r.vehicle.change_availability(True)
                print("Vehicle returned successfully")
                return

        print("No active rental found")

    def show_rentals(self):
        for r in self.rentals:
            print("--------------------")
            r.display_rental()

    def total_rental_amount(self):
        total=0
        for r in self.rentals:
            total=total+r.amount
        return total


# 8. CREATE OBJECTS

agency=RentalAgency()

car1=Car(101,"Toyota","Innova",2000,7)
bike1=Bike(102,"Honda","Shine",800,125)
premium1=PremiumCar(103,"BMW","5 Series",5000,5,"Luxury Interior")

agency.add_vehicle(car1)
agency.add_vehicle(bike1)
agency.add_vehicle(premium1)

c1=Customer(1,"Nazeer","8976543210","DL123")
agency.add_customer(c1)


# 9. MENU

while True:
    print("\n--- DRIVE EASY RENTALS ---")
    print("1. Add Car")
    print("2. Register Customer")
    print("3. Show Vehicles")
    print("4. Rent Vehicle")
    print("5. Return Vehicle")
    print("6. Show Rental History")
    print("7. Demonstrate Polymorphism")
    print("8. Total Rental Amount")
    print("9. Exit")

    try:
        choice=int(input("Enter choice: "))

        if choice==1:
            vid=int(input("Vehicle ID: "))
            brand=input("Brand: ")
            model=input("Model: ")
            rate=float(input("Rate per day: "))
            seats=int(input("Number of seats: "))

            car=Car(vid,brand,model,rate,seats)
            agency.add_vehicle(car)

        elif choice==2:
            cid=int(input("Customer ID: "))
            name=input("Name: ")
            phone=input("Phone: ")
            licence=input("Licence number: ")

            customer=Customer(cid,name,phone,licence)
            agency.add_customer(customer)

        elif choice==3:
            agency.show_vehicles()

        elif choice==4:
            cid=int(input("Customer ID: "))
            vid=int(input("Vehicle ID: "))
            days=int(input("Rental days: "))

            agency.rent_vehicle(cid,vid,days)

        elif choice==5:
            vid=int(input("Vehicle ID: "))
            agency.return_vehicle(vid)

        elif choice==6:
            agency.show_rentals()

        elif choice==7:
            for v in agency.vehicles:
                print(v.brand,v.calculate_cost(2))

        elif choice==8:
            print("Total rental amount:",agency.total_rental_amount())

        elif choice==9:
            print("Thank you")
            break

        else:
            print("Invalid choice")

    except ValueError:
        print("Please enter a valid number")
```
